## Exercise 3: Create a New Column

Create a new column called `average_score`.

The average score should be calculated from:

- `math_score`
- `english_score`
- `science_score`


```python
# Your code here
students["average_score"] = (students["math_score"] +
                              students["english_score"] +
                              students["science_score"]) / 3

print(students)
print(f" {type(students['average_score'])}")
```

## Exercise 4: Add a Pass / Fail Column

Create a new column called `status`.

If the student's `average_score` is **70 or more**, the status should be `"Pass"`.

Otherwise, the status should be `"Fail"`.


```python
# Your code here
import numpy as np
students["status"] = students["average_score"].apply(lambda x: "Pass" if x >= 70 else "Fail")
students
```

## Exercise 12: Challenge

Create a column called `performance_level`.

Rules:

- If `average_score` is 90 or more → `"Excellent"`
- If `average_score` is 80 to 89 → `"Very Good"`
- If `average_score` is 70 to 79 → `"Good"`
- Otherwise → `"Needs Improvement"`

Then display:

1. Student name
2. Average score
3. Performance level


```python
# Your code here
def assign_performance(score):
    if score >= 90:
        return "Excellent"
    elif score >= 80:
        return "Very Good"
    elif score >= 70:
        return "Good"
    else:
        return "Needs Improvement"

students["performance_level"] = students["average_score"].apply(assign_performance)

```

```python
students[["student_name", "average_score", "performance_level"]]
```
